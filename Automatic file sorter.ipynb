```python

import os
import shutil

# Set the path of the folder you want to organize
source_folder = r"C:\Users\subed\OneDrive\Desktop\everything"

# Define the categories to sort files into
categories = {
    'Images': ['.jpg', '.jpeg', '.png', '.gif', '.bmp', '.tiff'],
    'Text Files': ['.txt', '.md'],
    'PDFs': ['.pdf'],
    'Documents': ['.docx', '.doc', '.pptx', '.xlsx'],
    'Archives': ['.zip', '.rar', '.tar', '.gz'],
    'Audio': ['.mp3', '.wav', '.flac'],
    'Videos': ['.mp4', '.avi', '.mov', '.mkv']
}

# Create the categories directories if they don't exist
for category in categories.keys():
    category_folder = os.path.join(source_folder, category)
    if not os.path.exists(category_folder):
        os.makedirs(category_folder)

# Function to sort files based on their extension
def sort_files():
    for file_name in os.listdir(source_folder):
        file_path = os.path.join(source_folder, file_name)
        
        # Skip directories
        if os.path.isdir(file_path):
            continue
        
        file_extension = os.path.splitext(file_name)[1].lower()

        # Find the correct category for the file
        for category, extensions in categories.items():
            if file_extension in extensions:
                # Move the file to the corresponding category folder
                category_folder = os.path.join(source_folder, category)
                shutil.move(file_path, os.path.join(category_folder, file_name))
                print(f"Moved {file_name} to {category}")
                break

# Call the sort_files function to organize files
sort_files()

print("File sorting complete!")
```
